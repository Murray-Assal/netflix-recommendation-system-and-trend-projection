```python
filename = "../Dataset.csv"
raw_df = pd.read_csv(filename)
```

```python
raw_df = raw_df[raw_df['country'] != 'Not Given']
raw_df = raw_df[raw_df['director'] != 'Not Given']
```

```python
raw_df['date_added'] = pd.to_datetime(raw_df['date_added'], errors='coerce')
raw_df["year_added"] = raw_df["date_added"].dt.year
raw_df["month_added"] = raw_df["date_added"].dt.month
raw_df["director_list"] = raw_df["director"].apply(list)
raw_df['release_year'] = pd.to_datetime(raw_df['release_year'], format='%Y', errors='coerce').dt.year
raw_df['listed_in'] = raw_df['listed_in'].str.split(',')
raw_df["genres"] = raw_df["listed_in"].apply(lambda a: [g.strip() for g in a])
```

```python
raw_df.to_parquet("../prepared_data.parquet", engine='pyarrow', compression=None, index=False)
```
